```python
#prediction
from sklearn.linear_model import LinearRegression
X =[[1],[2],[3],[4],[5]]

y = [30,60,75,95,100]

model = LinearRegression()

model.fit(X,y)

hours = float(input("Enter how many hours you studies:"))

result = model.predict([[hours]])
print(f"based on your {hours},you may score {result} ")



```
